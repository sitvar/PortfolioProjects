```python
split_cols = df['Blood Pressure'].str.split('/', expand=True)
split_cols.columns = ['Systolic', 'Diastolic']

split_cols['Systolic'] = pd.to_numeric(split_cols['Systolic'])
split_cols['Diastolic'] = pd.to_numeric(split_cols['Diastolic'])

bp_index = df.columns.get_loc('Blood Pressure')
for i, col in enumerate(split_cols.columns):
    df.insert(bp_index + 1 + i, col, split_cols[col])
df
```

```python
df = df.drop(columns = "Blood Pressure")
df
```

```python
df.isnull().sum()
```

```python
for column in df.columns:
    if set(df[column].unique()) == {0, 1}:
        df[column] = df[column].replace({0: 'No', 1: 'Yes'})
df
```

```python
import seaborn as sns
plt.figure(figsize=(12, 8))
sns.countplot(data=df, x='Country', hue='Heart Attack Risk', palette='Set2')
plt.title('Country vs Heart Attack Risk')
plt.xlabel('Country')
plt.ylabel('Count')
plt.xticks(rotation=90)
plt.legend(title='Heart Attack Risk', labels=['No', 'Yes'])
plt.show()
```

```python
risk_by_country = df.groupby(['Country', 'Heart Attack Risk']).size().unstack(fill_value=0)
risk_percentages_by_country = risk_by_country.div(risk_by_country.sum(axis=1), axis=0) * 100
for country in risk_percentages_by_country.index:
    plt.figure(figsize=(6, 6))
    plt.pie(risk_percentages_by_country.loc[country], labels=['No Risk', 'At Risk'], autopct='%1.1f%%', startangle=90, colors=['#034748', '#CE4760'])
    plt.title(f'Heart Attack Risk Distribution in {country}')
    plt.show()
```
